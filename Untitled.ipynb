```python
dataset_id = 'c3s_obs-oc_glo_bgc-plankton_my_l4-multi-4km_P1M'
```

```python
ds = copernicusmarine.open_dataset(dataset_id)
```

```python
ds
```

```python
%%time
da = ds['CHL'].sel(time='2015-10-01', method='nearest').load() 
```

```python
min_val = da.min().item()
max_val = da.max().item()

print("Min CHL:", min_val)
print("Max CHL:", max_val)
```

```python
da = ds['CHL'].sel(
    time='2015-10-01',
    longitude=slice(-180, -70),
    latitude=slice(-25, 25)
).load()
```

```python
print("Min:", da.min().item())
print("Max:", da.max().item())
```

```python
ds['CHL'].attrs.get('units')
```

```python
import hvplot.xarray
import holoviews as hv

hv.extension('bokeh')

da.hvplot.quadmesh(
    x='longitude',
    y='latitude',
    geo=True,
    rasterize=True,
    cmap='turbo',
    tiles='OSM',
    frame_width=500
)
```

```python
import hvplot.xarray
import holoviews as hv

hv.extension('bokeh')

# --- subset region ---
da_reg = da.sel(
    longitude=slice(-180, -70),
    latitude=slice(-25, 25)
)

# --- get unit safely ---
unit = da.attrs.get('units', 'unknown unit')

# --- plot ---
da_reg.hvplot.quadmesh(
    x='longitude',
    y='latitude',
    geo=True,
    rasterize=True,
    cmap='viridis',
    tiles='OSM',
    clim=(0.012218493036925793, 60.729652404785156),
    frame_width=500,
    colorbar=True,
    clabel=f"CHL ({unit})",
    title="Chlorophyll (2015-10 subset)"
)
```

```python
import hvplot.xarray
import holoviews as hv

hv.extension('bokeh')

# --- subset region ---
da_reg = da.sel(
    longitude=slice(-180, -70),
    latitude=slice(-25, 25)
)

# --- remove missing values (masking) ---
da_clean = da_reg.where(da_reg.notnull())

# --- unit ---
unit = da.attrs.get('units', 'unknown unit')

# --- plot ---
da_clean.hvplot.quadmesh(
    x='longitude',
    y='latitude',
    geo=True,
    rasterize=True,
    cmap='viridis',
    tiles='OSM',
    clim=(0.012218493036925793, 60.729652404785156),
    frame_width=500,
    colorbar=True,
    clabel=f"CHL ({unit})",
    title="Chlorophyll (2015-10 subset)"
)
```
